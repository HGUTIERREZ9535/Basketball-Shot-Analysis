```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,5))

for contested in [False, True]:
    data = distance_contested[
        distance_contested.index.get_level_values("contested") == contested
    ]
    
    plt.plot(
        data.index.get_level_values("shot_distance").astype(str),
        data["Make Percentage"],
        marker="o",
        label=str(contested)
    )

plt.xlabel("Shot Distance")
plt.ylabel("Make Percentage")
plt.title("Make Percentage by Shot Distance and Contest Status")
plt.legend(title="Contested")
plt.savefig("Distance_Contested.png", bbox_inches="tight")

plt.show()
```
